### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/student-mat.csv).

### Step 3. Assign it to a variable called df.

```python
df = pd.read_csv("student_alcohol.csv")
df
```

### Step 4. For the purpose of this exercise slice the dataframe from 'school' until the 'guardian' column

```python
stud_alcoh = df.loc[:, "school":"guardian"]
stud_alcoh.head()
```

### Step 5. Create a lambda function that will capitalize strings.

```python
capitalize = lambda value: value.capitalize() if isinstance(value, str) else value
```

### Step 6. Capitalize both Mjob and Fjob

```python
stud_alcoh["Mjob"].apply(capitalize)
stud_alcoh["Fjob"].apply(capitalize)
```

### Step 7. Print the last elements of the data set.

```python
stud_alcoh.tail()
```

### Step 8. Did you notice the original dataframe is still lowercase? Why is that? Fix it and capitalize Mjob and Fjob.

```python
stud_alcoh[["Mjob", "Fjob"]] = stud_alcoh[["Mjob", "Fjob"]].apply(lambda col: col.apply(capitalize))
stud_alcoh.tail()
```

### Step 9. Create a function called majority that returns a boolean value to a new column called legal_drinker (Consider majority as older than 17 years old)

```python
def majority(age):
    return age > 17


stud_alcoh["legal_drinker"] = stud_alcoh["age"].apply(majority)
stud_alcoh[["age", "legal_drinker"]].head()
```

```python
stud_alcoh[["Mjob", "Fjob", "legal_drinker"]].tail()
```

### Step 10. Multiply every number of the dataset by 10.
