### Step 1. Import the necessary libraries:

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/tips.csv).

### Step 3. Assign it to a variable called tips

```python
tips = pd.read_csv("tips.csv")
tips
```

### Step 4. Delete the Unnamed 0 column

```python
tips = tips.drop(columns="Unnamed: 0")
tips.head()
```

### Step 5. Plot the total_bill column histogram

```python
tips["total_bill"].plot(kind="hist", bins=20)
plt.xlabel("Total Bill")
plt.title("Distribution of Total Bill")
plt.tight_layout()
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
tips.plot(kind="scatter", x="total_bill", y="tip", alpha=0.6)
plt.title("Total Bill vs Tip")
plt.tight_layout()
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.pairplot(tips[["total_bill", "tip", "size"]])
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill by Day")
plt.tight_layout()
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.stripplot(data=tips, x="tip", y="day", hue="sex", dodge=True, alpha=0.6)
plt.title("Tip by Day and Sex")
plt.tight_layout()
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(data=tips, x="day", y="total_bill", hue="time")
plt.title("Total Bill by Day and Time")
plt.tight_layout()
plt.show()
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
sns.histplot(tips.loc[tips["time"] == "Dinner", "tip"], bins=15, ax=axes[0])
axes[0].set_title("Dinner Tips")
sns.histplot(tips.loc[tips["time"] == "Lunch", "tip"], bins=15, ax=axes[1])
axes[1].set_title("Lunch Tips")
plt.tight_layout()
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharey=True)
for ax, sex in zip(axes, ["Male", "Female"]):
    subset = tips.loc[tips["sex"] == sex]
    sns.scatterplot(data=subset, x="total_bill", y="tip", hue="smoker", ax=ax)
    ax.set_title(sex)

plt.tight_layout()
plt.show()
```
