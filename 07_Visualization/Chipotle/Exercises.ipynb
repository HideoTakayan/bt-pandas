### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# set this so the graphs open internally
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv).

### Step 3. Assign it to a variable called chipo.

```python
chipo = pd.read_csv("https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/chipotle.tsv", sep="\t")
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).str.strip().astype(float)
chipo
```

### Step 4. See the first 10 entries

```python
chipo.head(10)
```

### Step 5. Create a histogram of the top 5 items bought

```python
top_items = chipo.groupby("item_name")["quantity"].sum().sort_values(ascending=False).head(5)
top_items.plot(kind="bar")
plt.xlabel("Item")
plt.ylabel("Quantity Ordered")
plt.title("Top 5 Most Ordered Chipotle Items")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Step 6. Create a scatterplot with the number of items orderered per order price
#### Hint: Price should be in the X-axis and Items ordered in the Y-axis

```python
order_stats = (
    chipo.assign(line_total=chipo["quantity"] * chipo["item_price"])
    .groupby("order_id")
    .agg(items_ordered=("quantity", "sum"), order_price=("line_total", "sum"))
)

plt.scatter(order_stats["order_price"], order_stats["items_ordered"], alpha=0.6)
plt.xlabel("Order Price")
plt.ylabel("Items Ordered")
plt.title("Items Ordered Per Order Price")
plt.tight_layout()
plt.show()
```

### Step 7. BONUS: Create a question and a graph to answer your own question.

```python
revenue_by_item = (
    chipo.assign(revenue=chipo["quantity"] * chipo["item_price"])
    .groupby("item_name")["revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

revenue_by_item.plot(kind="bar", color="tomato")
plt.xlabel("Item")
plt.ylabel("Revenue")
plt.title("Top 10 Items by Revenue")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
