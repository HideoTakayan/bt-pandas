### Step 1. Import the necessary libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/thieu1995/csv-files/main/data/pandas/titanic_train.csv)

### Step 3. Assign it to a variable titanic 

```python
titanic = pd.read_csv("titanic.csv")
titanic
```

### Step 4. Set PassengerId as the index 

```python
titanic = titanic.set_index("PassengerId")
titanic.head()
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
sex_counts = titanic["Sex"].value_counts()
sex_counts.plot(kind="pie", autopct="%1.1f%%", ylabel="")
plt.title("Male/Female Proportion")
plt.show()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
colors = titanic["Sex"].map({"male": "steelblue", "female": "salmon"})
plt.scatter(titanic["Fare"], titanic["Age"], c=colors, alpha=0.6)
plt.xlabel("Fare")
plt.ylabel("Age")
plt.title("Fare vs Age by Gender")
plt.tight_layout()
plt.show()
```

### Step 7. How many people survived?

```python
titanic["Survived"].sum()
```

### Step 8. Create a histogram with the Fare payed

```python
titanic["Fare"].plot(kind="hist", bins=30)
plt.xlabel("Fare")
plt.title("Fare Distribution")
plt.tight_layout()
plt.show()
```

### BONUS: Create your own question and answer it.

```python
survival_by_class = titanic.groupby("Pclass")["Survived"].mean()
survival_by_class.plot(kind="bar", color="teal")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Passenger Class")
plt.tight_layout()
plt.show()
```
